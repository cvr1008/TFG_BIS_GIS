Nueva propuesta objetivos:

Diferenciar entre variables clínicas de alta frecuencia, observaciones
puntuales, cambios de tratamiento y eventos clínicos, evitando representar
como mediciones continuas aquellos datos que solo se registran de forma
periódica o manual.

Permitir la consulta de todos los episodios y sesiones BIS asociados a un
mismo paciente, indicando para cada sesión si existe información ICCA
temporalmente compatible.

| Extensión | Contenido |
|:---|:---|
| `.spa` | Variables procesadas por el monitor y marcas temporales. |
| `.f_a` | Matriz espectral calculada por el monitor. |
| `.r2a` | Señal EEG cruda de una monitorización unilateral. |
| `.r4a` | Señal EEG cruda de una monitorización bilateral. |
| `.h_a` | Cabecera con los parámetros necesarios para interpretar el EEG crudo. |
| `.t_a` | Información temporal asociada al comienzo de la señal cruda. |
